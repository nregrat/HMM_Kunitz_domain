# Model Traning

The `hmmbuild` command from the `HMMER` (version 3.4) suite was used to construct a the HMM from the multiple sequence alignment. In this specific case, `kunitz_model_mtm.hmm` was the output HMM file, and `filter_mtm_structural_alg.fasta` served as the input multiple sequence alignment in FASTA format. This input alignment was generated from the structural alignment of Kunitz domains.
```bash
hmmbuild --amino kunitz_model_mtm.hmm filter_mtm_structural_alg.fasta
```
